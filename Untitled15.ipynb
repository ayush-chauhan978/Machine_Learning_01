```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\USER\OneDrive\Documents\customer.csv")
df
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=['Customer_Type']),df['Customer_Type'],
                                                       test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['Age']),
    ('tnf2',OrdinalEncoder(categories=[['Poor','Average','Good','Excellent']]),['Reviews']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['Gender','Education'])
],remainder='passthrough')
```

```python
transformer.fit_transform(x_train).shape
```

```python
transformer.transform(x_test).shape
```

```python
# fit ONLY on training data
transformer.fit(x_train)

# transform both using same learned structure
x_train_transformed = transformer.transform(x_train)
x_test_transformed = transformer.transform(x_test)

print(x_train_transformed.shape)
print(x_test_transformed.shape)
```
