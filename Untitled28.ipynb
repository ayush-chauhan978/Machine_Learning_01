```python
x=df
y=df['SalePrice']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
x_train['garage_imputed']=x_train['Garage_Quality']
x_test['garage_imputed']=x_test['Garage_Quality']

x_train['fire_imputed'] = x_train['Fireplace_Quality']
x_test['fire_imputed']= x_test['Fireplace_Quality']
```

```python
x_train
```

```python
x_train['garage_imputed'][x_train['garage_imputed'].isnull()] = x_train['Garage_Quality'].dropna().sample(x_train['Garage_Quality'].isnull().sum()).values
x_test['garage_imputed'][x_test['garage_imputed'].isnull()] = x_train['Garage_Quality'].dropna().sample(x_test['Garage_Quality'].isnull().sum()).values

x_train['fire_imputed'][x_train['fire_imputed'].isnull()]= x_train['Fireplace_Quality'].dropna().sample(x_train['Fireplace_Quality'].isnull().sum()).values
x_test['fire_imputed'][x_test['fire_imputed'].isnull()]= x_train['Fireplace_Quality'].dropna().sample(x_test['Fireplace_Quality'].isnull().sum()).values
```

```python
temp = pd.concat([
    x_train['Garage_Quality'].value_counts()/len(x_train['Garage_Quality'].dropna()),
    x_train['garage_imputed'].value_counts()/len(x_train)
],
                 axis=1)

temp.columns=['original','imputed']
```
