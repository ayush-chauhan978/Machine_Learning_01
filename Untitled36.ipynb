```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\USER\OneDrive\Documents\titanic.csv")[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
x = df.iloc[:,0:4]
y= df.iloc[:,-1]
```

```python
x
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy',cv=5))
```

```python
# featue construction
x['family_size']= x['SibSp'] + x['Parch'] +1
```

```python
x
```

```python
def myfunc(num):
    if num ==1:
        return 0 # alone
    elif num>1 and num <=4:
        return 1 # small family
    else:
        return 2 # large family
```

```python
x['family_type'] = x['family_size'].apply(myfunc)
```

```python
x
```

```python
x.drop(columns=['SibSp','Parch','family_size'],inplace = True)
```

```python
x
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy',cv=5))
```
