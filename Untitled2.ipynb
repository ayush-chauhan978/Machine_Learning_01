```python
import os
from PIL import Image

path = '/content/kagglecatsanddogs_3367a/PetImages'

for folder in ['Cat', 'Dog']:
    folder_path = os.path.join(path, folder)

    for file in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file)

        try:
            img = Image.open(file_path)
            img.verify()  # check corruption
        except:
            os.remove(file_path)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,Flatten
from keras.applications.vgg16 import VGG16
```

```python
conv_base=VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(150,150,3)
)
```

```python
model = Sequential()

model.add(conv_base)
model.add(Flatten())
model.add(Dense(256,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
conv_base.trainable = False
```

```python
train_ds = keras.utils.image_dataset_from_directory(
    directory = '/content/kagglecatsanddogs_3367a/PetImages',
    labels = 'inferred',
    subset='training',
    validation_split=0.2,
    seed=123,
    label_mode='int',
    batch_size=32,
    image_size=(150,150)
)

val_ds = keras.utils.image_dataset_from_directory(
    directory = '/content/kagglecatsanddogs_3367a/PetImages',
    labels='inferred',
    label_mode='int',
    subset='validation',
    validation_split=0.2,
    seed=123,
    batch_size=32,
    image_size=(150,150)
)
```

```python
def process(image,label):
  image = tensorflow.cast(image/255. ,tensorflow.float32)
  return image,label

train_ds = train_ds.map(process)
val_ds = val_ds.map(process)
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
train_ds = train_ds.apply(tensorflow.data.experimental.ignore_errors())
val_ds = val_ds.apply(tensorflow.data.experimental.ignore_errors())

history = model.fit(train_ds, epochs=5, validation_data=val_ds)
```
