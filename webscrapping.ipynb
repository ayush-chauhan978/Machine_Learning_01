```python
final_ =[]
for j in range(1,3):
    url='https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={}'.format(j)
    headers={'user-agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'}
    webpage = requests.get(url,headers=headers).text
    soup= BeautifulSoup(webpage,'lxml')
    com=soup.find_all('div',class_='companyCardWrapper')
    
    name=[]
    rating = []
    reviews=[]
    description = []
    for i in com:
        name.append(i.find('h2').text.strip())
        rating.append(i.find_all('div',class_='rating_star_container')[0].text.strip())
        reviews.append(i.find_all('a',class_='companyCardWrapper__ActionWrapper')[0].text.strip())
        description.append(i.find_all('span',class_='companyCardWrapper__interLinking')[0].text.strip())
    d={'NAME':name,'RATINGS':rating,'REVIEWS':reviews,'DESCRIPTION':description}
    df=pd.DataFrame(d)
    final_.append(df)
final_list=pd.concat(final_,ignore_index=True)
final_list
```
