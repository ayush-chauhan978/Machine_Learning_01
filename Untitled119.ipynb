```python
faqs='''What is machine learning ? Machine learning is a field of artificial intelligence that allows systems to learn from data .

What is deep learning ? Deep learning is a subset of machine learning that uses neural networks with many layers .

What is artificial intelligence ? Artificial intelligence is the simulation of human intelligence in machines .

What is a neural network ? A neural network is a system of algorithms inspired by the human brain .

What is supervised learning ? Supervised learning is a type of machine learning where the model is trained on labeled data .

What is unsupervised learning ? Unsupervised learning is a type of learning where the model finds patterns in unlabeled data .

What is overfitting ? Overfitting occurs when a model learns the training data too well and performs poorly on new data .

What is underfitting ? Underfitting occurs when a model is too simple to capture the underlying pattern of the data .

What is a dataset ? A dataset is a collection of data used for training and testing models .

What is a feature ? A feature is an individual measurable property or characteristic of a phenomenon .

What is a label ? A label is the output or target variable in supervised learning .

What is gradient descent ? Gradient descent is an optimization algorithm used to minimize loss functions .

What is a loss function ? A loss function measures how well a model's predictions match the actual values .

What is training data ? Training data is the data used to train a machine learning model .

What is testing data ? Testing data is used to evaluate the performance of a trained model .

What is validation data ? Validation data is used to tune model parameters during training .

What is natural language processing ? Natural language processing is a field of AI that focuses on interaction between computers and human language .

What is tokenization ? Tokenization is the process of breaking text into smaller units like words or sentences .

What is stemming ? Stemming is the process of reducing words to their root form .

What is lemmatization ? Lemmatization is the process of converting words to their base form .

What is a language model ? A language model predicts the probability of a sequence of words .

What is next word prediction ? Next word prediction is the task of predicting the next word in a sequence of text .

What is perplexity ? Perplexity is a metric used to evaluate language models .

What is a corpus ? A corpus is a large collection of text used for training language models .'''
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts([faqs])
```

```python
len(tokenizer.word_index)
```

```python
input_sequences=[]
for sentence in faqs.split('\n'):
    tokenized_sentence =tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1, len(tokenized_sentence)):
        input_sequences.append(tokenized_sentence[:i+1])
```

```python
max_len=max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
pad_input_sequences=pad_sequences(input_sequences ,maxlen = max_len,padding='pre')
```

```python
x = pad_input_sequences[:,:-1]
```

```python
y = pad_input_sequences[:,-1]
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=141)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
model =Sequential()
model.add(Embedding(141,100,input_length=21))
model.add(LSTM(150))
model.add(Dense(141,activation='softmax'))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x,y,epochs=100)
```

```python
text_token=tokenizer.texts_to_sequences([text])[0]
```

```python
pad_token=pad_sequences([text_token],maxlen=21,padding='pre')
```

```python
model.predict(pad_token).shape
```

```python
import numpy as np
```

```python
pos=np.argmax(model.predict(pad_token))
```

```python
for word,index in tokenizer.word_index.items():
    if index==pos:
        print(word)
```

```python
import time 
text='Machine'
for i in range(10):
    text_token=tokenizer.texts_to_sequences([text])[0]
    pad_token=pad_sequences([text_token],maxlen=21,padding='pre')
    pos=np.argmax(model.predict(pad_token))

    for word,index in tokenizer.word_index.items():
        if index==pos:
            text = text+" " + word
            print(text)
            time.sleep(2)
```
