```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
def analyzer(max_depth):
    data = pd.read_csv(r"C:\Users\USER\OneDrive\Documents\social.csv")

    x = data.iloc[:, 2:4].values   # Age, Salary
    y = data.iloc[:, -1].values    # Purchased

    clf = DecisionTreeClassifier(max_depth=max_depth)
    clf.fit(x, y)

    # Create meshgrid
    a = np.arange(start=x[:,0].min()-1, stop=x[:,0].max()+1, step=0.1)
    b = np.arange(start=x[:,1].min()-1, stop=x[:,1].max()+1, step=1000)

    xx, yy = np.meshgrid(a, b)

    input_array = np.c_[xx.ravel(), yy.ravel()]

    labels = clf.predict(input_array)

    # Plot decision boundary
    plt.contourf(xx, yy, labels.reshape(xx.shape), alpha=0.5)

    # Plot actual data
    plt.scatter(x[:,0], x[:,1], c=y)

    plt.xlabel("Age")
    plt.ylabel("Estimated Salary")
    plt.title(f"Decision Tree (max_depth={max_depth})")

    plt.show()
```

```python
analyzer(max_depth=8) # max_depth means only one splitting
```
