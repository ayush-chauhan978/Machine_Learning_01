```python
df['Travel_With'].unique()
```

```python
df['number_numerical']=pd.to_numeric(df['Travel_With'],errors='coerce',downcast='integer')
```

```python
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['Travel_With'],np.nan)
```

```python
df["Ticket"].unique()
```

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)')# captures numeical part
df['cabin_cat'] = df['Cabin'].str[0]# capture the first letter 
```

```python
df.head()
```

```python
df['tic_num'] = df['Ticket'].str.extract('(\d+)')
df['tic_cat'] = df['Ticket'].str.extract('([A-Za-z]+)')
df['tic_cat'] = df['tic_cat'].fillna("num")
```
